```python
import torchsummary 
import torch.nn as nn 

from torchvision.models import resnet18, ResNet18_Weights

class ResNetTest(nn.Module):
    def __init__(self):
        super().__init__()
        self.weights = ResNet18_Weights.IMAGENET1K_V1
        self.model = resnet18(weights=self.weights)

    def forward(self, x):
        return self.model(x)

device = 'cuda:0'
model = ResNetTest().to(device)
torchsummary.summary(model, input_size=(3, 140, 162), batch_size=32)
```
